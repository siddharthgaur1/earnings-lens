# Correlation analysis -- NLP features vs financial outcomes

Runs `analysis/correlations.py` for real against the synthetic corpus's 30 transcripts and their synthetic financial outcomes. Spearman rank correlation + p-value for every (feature, outcome) pair -- non-significant results are reported too, not filtered out.

```python
import sys

sys.path.insert(0, '.')
import pandas as pd

from analysis.correlations import run

pd.set_option('display.max_rows', None)
result = run()
result
```

## Significant pairs only (alpha=0.05)

```python
result[result['significant'] == True].sort_values('p_value')
```

## Summary

```python
n_sig = int(result['significant'].sum())
print(f'{n_sig} of {len(result)} feature-outcome pairs significant at alpha=0.05')
print('Strongest by |rho| among significant pairs:')
sig = result[result['significant'] == True]
if len(sig):
    print(sig.reindex(sig['rho'].abs().sort_values(ascending=False).index).iloc[0])
else:
    print('(none significant)')
```

## Interpretation

This is a **synthetic dataset** with 30 observations, so p-values here demonstrate the *method* (Spearman correlation + honest significance reporting), not a real market finding. With n=30 and alpha=0.05, a handful of pairs are expected to cross significance by chance alone even under a true null -- treat the specific significant pairs above as illustrative, not as investment signal.
